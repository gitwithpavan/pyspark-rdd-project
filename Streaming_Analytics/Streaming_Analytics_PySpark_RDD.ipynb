# PySpark RDD Data Engineering Project — Streaming Analytics

This project uses two related input files and focuses on practical PySpark RDD transformations, joins, aggregations, filtering, sorting, deduplication, and business analysis.

**Input files:** `users.csv`, `sessions.csv`

**Total scenario questions:** 30

```python
from pyspark import SparkContext
import csv

sc = SparkContext.getOrCreate()

rdd1 = sc.textFile("users.csv").mapPartitions(lambda it: csv.reader(it)).filter(lambda x: x[0]!="user_id")
rdd2 = sc.textFile("sessions.csv").mapPartitions(lambda it: csv.reader(it)).filter(lambda x: x[0]!="session_id")
```

## Q1. Record Profiling

Count records in both input files and inspect a sample.

```python
print(rdd1.count(), rdd2.count()); print(rdd1.take(5)); print(rdd2.take(5))
```

## Q2. Key-Based Join

Join the two datasets using user_id and user_id.

```python
a=rdd1.map(lambda x:(x[0],x)); b=rdd2.map(lambda x:(x[1],x)); print(a.join(b).take(20))
```

```python
print(rdd1.map(lambda x:(x[2],1)).reduceByKey(lambda a,b:a+b).collect())
```

## Q7. Numeric Average

Calculate average watch_minutes.

```python
r=rdd2.map(lambda x:(float(x[4]),1)).reduce(lambda a,b:(a[0]+b[0],a[1]+b[1])); print(r[0]/r[1])
```

## Q8. Category-Wise Total

Calculate total watch_minutes by genre.

```python
print(rdd2.map(lambda x:(x[3],float(x[4]))).reduceByKey(lambda a,b:a+b).sortBy(lambda x:x[1],ascending=False).collect())
```

```python
r=rdd2.map(lambda x:(x[3],(float(x[4]),1))).reduceByKey(lambda a,b:(a[0]+b[0],a[1]+b[1])).mapValues(lambda x:x[0]/x[1]); print(r.collect())
```

## Q10. Top 10 Values

Identify duplicate user_id values.

```python
print(rdd2.map(lambda x:(x[1],1)).reduceByKey(lambda a,b:a+b).filter(lambda x:x[1]>1).collect())
```

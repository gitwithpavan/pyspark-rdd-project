# PySpark RDD Data Engineering Project — Ecommerce Order Analytics

This project uses two related input files and focuses on practical PySpark RDD transformations, joins, aggregations, filtering, sorting, deduplication, and business analysis.

**Input files:** `customers.csv`, `orders.csv`

**Total scenario questions:** 30

```python
from pyspark import SparkContext
import csv

sc = SparkContext.getOrCreate()

rdd1 = sc.textFile("customers.csv").mapPartitions(lambda it: csv.reader(it)).filter(lambda x: x[0]!="customer_id")
rdd2 = sc.textFile("orders.csv").mapPartitions(lambda it: csv.reader(it)).filter(lambda x: x[0]!="order_id")
```

## Q1. Record Profiling

Count records in both input files and inspect a sample.

```python
print(rdd1.count(), rdd2.count()); print(rdd1.take(5)); print(rdd2.take(5))
```

## Q2. Key-Based Join

Join the two datasets using customer_id and customer_id.

```python
a=rdd1.map(lambda x:(x[0],x)); b=rdd2.map(lambda x:(x[1],x)); print(a.join(b).take(20))
```

## Q3. Group Count

Count records by city.

```python
print(rdd1.map(lambda x:(x[2],1)).reduceByKey(lambda a,b:a+b).collect())
```

## Q7. Numeric Average

Calculate average order_value.

```python
r=rdd2.map(lambda x:(float(x[5]),1)).reduce(lambda a,b:(a[0]+b[0],a[1]+b[1])); print(r[0]/r[1])
```

## Q8. Category-Wise Total

Calculate total order_value by category.

```python
print(rdd2.map(lambda x:(x[2],float(x[5]))).reduceByKey(lambda a,b:a+b).sortBy(lambda x:x[1],ascending=False).collect())
```

## Q9. Category-Wise Average

Calculate average order_value by category.

```python
r=rdd2.map(lambda x:(x[2],(float(x[5]),1))).reduceByKey(lambda a,b:(a[0]+b[0],a[1]+b[1])).mapValues(lambda x:x[0]/x[1]); print(r.collect())
```

## Q10. Top 10 Values

Find the ten largest order_value values.

## Q11. Bottom 10 Values

Find the ten smallest order_value values.

```python
print(rdd2.map(lambda x:(x[5],float(x[5]))).sortBy(lambda x:x[1]).take(10))
```

## Q12. Duplicate Key Audit

Identify duplicate customer_id values.

```python
print(rdd2.map(lambda x:(x[1],1)).reduceByKey(lambda a,b:a+b).filter(lambda x:x[1]>1).collect())
```

## Q13. Filtered Numeric

Find records where order_value is above its median-like operational threshold.

```python
print(rdd2.filter(lambda x:float(x[5])>float(rdd2.map(lambda y:float(y[5])).mean())).take(30))
```

## Q14. Sorted Categories

Sort category totals by descending order_value.

## Q15. Join and Aggregate

Join the files and calculate total order_value by city.

```python
a=rdd1.map(lambda x:(x[0],x[2])); b=rdd2.map(lambda x:(x[1],float(x[5]))); print(b.join(a).map(lambda x:(x[1][1],x[1][0])).reduceByKey(lambda a,b:a+b).collect())
```
